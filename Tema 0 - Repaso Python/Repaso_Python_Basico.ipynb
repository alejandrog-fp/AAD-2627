## 1. Variables y tipos de datos

En Python **no se declara el tipo** de una variable: se crea en el momento en que se le asigna un
valor, y su tipo se deduce automáticamente (*tipado dinámico*).

> 🔶 **Diferencia con Java:** en Java hay que escribir `int edad = 20;` o `String nombre = "Ana";`.
> En Python basta con `edad = 20`. La variable puede además cambiar de tipo en cualquier momento
> (aunque, por claridad, es buena práctica no hacerlo).

Los tipos básicos son:

| Tipo | Ejemplo | Equivalente en Java |
|---|---|---|
| `int` | `edad = 20` | `int`, `long` |
| `float` | `precio = 19.99` | `double`, `float` |
| `str` | `nombre = "Ana"` | `String` |
| `bool` | `activo = True` | `boolean` |
| `NoneType` | `resultado = None` | `null` |


````python
edad = 20
precio = 19.99
nombre = "Ana"
activo = True
resultado = None

print(edad, type(edad))
print(precio, type(precio))
print(nombre, type(nombre))
print(activo, type(activo))
print(resultado, type(resultado))

# La misma variable puede cambiar de tipo (no es recomendable abusar de esto)
variable = 5
print(variable, type(variable))
variable = "ahora soy un texto"
print(variable, type(variable))
````

````python
# Conversión entre tipos (equivalente al "casting" de Java, pero mediante funciones)
texto_numero = "42"
numero = int(texto_numero)          # str -> int
decimal = float(numero)             # int -> float
texto_de_nuevo = str(decimal)       # float -> str

print(numero, type(numero))
print(decimal, type(decimal))
print(texto_de_nuevo, type(texto_de_nuevo))

# Cuidado: convertir un texto no numérico lanza una excepción (lo veremos en el apartado 12)
int("hola")   # -> ValueError: invalid literal for int() with base 10: 'hola'
````

## 2. Operadores

### 2.1. Aritméticos

| Operador | Significado | Ejemplo |
|---|---|---|
| `+` `-` `*` | suma, resta, multiplicación | `3 + 2` -> `5` |
| `/` | división (**siempre** devuelve `float`) | `7 / 2` -> `3.5` |
| `//` | división **entera** (redondeo hacia abajo) | `7 // 2` -> `3` |
| `%` | resto de la división (módulo) | `7 % 2` -> `1` |
| `**` | potencia | `2 ** 10` -> `1024` |

> 🔶 **Diferencia con Java:** en Java, `7 / 2` con dos `int` da `3` (división entera implícita). En
> Python, `/` **siempre** devuelve `float`; si se quiere división entera hay que usar `//` de forma
> explícita.


````python
a, b = 7, 2

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)    # división real -> float, aunque a y b sean int
print("a // b =", a // b)   # división entera
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

### 2.2. Comparación y lógicos

| Comparación | Significado | | Lógico | Significado |
|---|---|---|---|---|
| `==` | igual a | | `and` | Y lógico |
| `!=` | distinto de | | `or` | O lógico |
| `>` `<` `>=` `<=` | mayor, menor, ... | | `not` | negación |

> 🔶 **Diferencia con Java:** Python usa las palabras `and`, `or`, `not` en lugar de `&&`, `||`, `!`.


````python
x, y = 5, 10

print(x == y, x != y, x < y, x >= y)
print((x < y) and (y > 0))
print((x > y) or (y > 0))
print(not (x == y))

# Python permite ENCADENAR comparaciones de forma muy legible (esto NO existe en Java)
nota = 7
print(0 <= nota <= 10)   # equivale a: 0 <= nota and nota <= 10
````

### 2.3. Asignación aumentada

Igual que en Java: `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`.


````python
contador = 0
contador += 1 # contador = contado + 1
contador += 5
contador *= 2 # contador = contador * 2
print(contador)   # (0+1+5) * 2 = 12
````

## 3. Estructuras de control: `if` / `elif` / `else`

La sintaxis es muy parecida a Java, pero con dos diferencias clave:

- **No se usan llaves `{ }`**: el bloque de código se delimita mediante la **indentación** (sangrado).
  Una indentación incorrecta es un **error de sintaxis** en Python, no solo un problema de estilo.
- **No se usan paréntesis obligatorios** alrededor de la condición (aunque se pueden poner).
- En vez de `else if`, se escribe **`elif`**.

```python
if condicion1:
    # bloque si condicion1 es True
elif condicion2:
    # bloque si condicion2 es True
else:
    # bloque si ninguna condición anterior se cumplió
```


````python
nota = 6.5

if nota >= 9:
    calificacion = "Sobresaliente"
elif nota >= 7:
    calificacion = "Notable"
elif nota >= 5:
    calificacion = "Aprobado"
else:
    calificacion = "Suspenso"

print(f"Nota {nota} -> {calificacion}")
````

### Operador condicional (equivalente al ternario `? :` de Java)

Python no tiene el operador `?:`, pero permite escribir un `if`/`else` **en una sola línea**, en el
mismo orden en que se leería la frase:


````python
nota = 6.5

# equivalente a: String resultado = (nota >= 5) ? "Aprobado" : "Suspenso";  (en Java)
resultado = "Aprobado" if nota >= 5 else "Suspenso"
print(resultado)
````

## 4. Bucles

### 4.1. `for`

El `for` de Python **siempre itera sobre un iterable** (una secuencia de elementos): una lista, una
cadena, un rango de números... No existe el `for` clásico de Java con contador (`for (int i=0; ...)`),
aunque `range()` permite conseguir el mismo efecto.

> 🔶 **Diferencia con Java:** en Java, `for (int i = 0; i < 5; i++)`. En Python, `for i in range(5)`.


````python
# Iterar sobre un rango de números (como el for clásico de Java)
for i in range(5):            # 0, 1, 2, 3, 4  (el límite superior NO se incluye)
    print(i, end=" ")
print()

for i in range(2, 10, 2):     # inicio=2, fin=10 (excluido), paso=2
    print(i, end=" ")
print()

# Iterar directamente sobre los elementos de una secuencia (lo más habitual en Python)
frutas = ["manzana", "pera", "plátano"]
for fruta in frutas:
    print(fruta.capitalize())

# Si además necesitamos el índice, se usa enumerate() (evita tener que llevar un contador manual)
for indice, fruta in enumerate(frutas): # [ [0, 'manzana'] , [1, "pera"] , ... ]
    print(f"{indice}: {fruta}")
````

### 4.2. `while`

Se comporta exactamente igual que en Java.


````python
contador = 0
while contador < 5:
    print("Iteración", contador)
    contador += 1     # 🔶 en Python NO existen los operadores ++ ni --
````

### 4.3. `break`, `continue` y la cláusula `else` en bucles

`break` y `continue` funcionan igual que en Java. Pero Python tiene una particularidad que **no**
existe en Java: los bucles `for`/`while` pueden llevar un bloque `else`, que se ejecuta **solo si el
bucle terminó sin un `break`**. Es muy útil, por ejemplo, para saber si una búsqueda tuvo éxito.


````python
numeros = [4, 8, 15, 16, 23, 42]

# break: corta el bucle en cuanto encuentra lo que busca
for n in numeros:
    if n == 15:
        print("¡Encontrado el 15!")
        break
    print("Revisando", n)

print()

# continue: salta a la siguiente iteración sin ejecutar el resto del cuerpo
for n in numeros:
    if n % 2 != 0:
        continue         # nos saltamos los números impares
    print(n, "es par")

print()

# for...else: el "else" se ejecuta si el bucle terminó SIN pasar por un break
buscado = 99
for n in numeros:
    if n == buscado:
        print(f"{buscado} encontrado")
        break
else:
    print(f"{buscado} NO está en la lista")
````

## 5. Cadenas de texto (*strings*)

Los `str` de Python son **inmutables** (igual que los `String` de Java) y admiten indexación y
*slicing* (extraer subcadenas mediante rangos), algo que Java no ofrece de forma tan directa.


````python
texto = "Desarrollo de Aplicaciones Multiplataforma"

print(len(texto))          # longitud
print(texto[0])            # primer carácter (los índices empiezan en 0, como en Java)
print(texto[-1])           # 🔶 índice NEGATIO: el último carácter (no existe en Java)
print(texto[0:11])         # slicing: caracteres del índice 0 al 10 (el 11 no se incluye)
print(texto[:11])          # equivalente: si se omite el inicio, se asume 0
print(texto[12:])          # desde el índice 12 hasta el final
print(texto[::-1][:20])    # slicing con paso -1 -> invierte la cadena (mostramos solo 20 caracteres)
````

````python
texto = "  Hola Mundo  "

print(repr(texto.strip()))       # elimina espacios al principio y al final
print(texto.strip().upper())     # mayúsculas
print(texto.strip().lower())     # minúsculas
print(texto.strip().replace("Mundo", "DAM"))
print(texto.strip().split(" "))  # divide la cadena en una lista de trozos
print("-".join(["A", "B", "C"])) # operación inversa: unir una lista en una cadena
print("Mundo" in texto)          # comprobar si una subcadena está contenida
print(texto.strip().startswith("Hola"))
````

### `f-strings`: la forma moderna de formatear cadenas

Equivalen a `String.format()` o a la concatenación con `+` de Java, pero de forma mucho más legible:
se escribe una `f` antes de las comillas y las variables van entre llaves `{ }`.


````python
nombre = "Ana"
nota = 8.567

# Antes de las f-strings (formas más antiguas, todavía se ven en código heredado)
print("Alumno: " + nombre + ", nota: " + str(nota))          # concatenación clásica
print("Alumno: {}, nota: {:.2f}".format(nombre, nota))        # .format()

# Forma moderna recomendada: f-string
print(f"Alumno: {nombre}, nota: {nota:.2f}")                  # :.2f -> 2 decimales
print(f"En mayúsculas: {nombre.upper()}")                     # se pueden llamar métodos dentro de {}
print(f"¿Aprobado? {'Sí' if nota >= 5 else 'No'}")             # incluso expresiones completas
````

## 6. Listas (`list`)

La lista es la estructura de datos más usada en Python. Es **mutable**, **ordenada** y puede contener
elementos de **cualquier tipo mezclados**.

> 🔶 **Diferencia con Java:** un `ArrayList<Integer>` de Java solo admite un tipo. Una lista de Python
> puede mezclar `int`, `str`, otras listas, etc. en la misma lista (aunque, por claridad de código, lo
> habitual es no mezclar tipos salvo que haya una buena razón).


````python
notas = [8.5, 6.0, 9.25, 5.5]

print(notas)
print(len(notas))
print(notas[0], notas[-1])       # acceso por índice (también admite índices negativos)
print(notas[1:3])                # slicing, igual que con las cadenas

notas.append(7.0)                # añadir al final
print("Tras append:", notas)

notas.insert(0, 10.0)            # insertar en una posición concreta
print("Tras insert:", notas)

notas.remove(6.0)                # elimina la PRIMERA aparición del valor indicado
print("Tras remove:", notas)

eliminada = notas.pop()          # elimina y devuelve el último elemento
print("Elemento eliminado con pop():", eliminada, "-> lista:", notas)

print("Ordenada:", sorted(notas))           # sorted() devuelve una nueva lista ordenada
notas.sort(reverse=True)                    # sort() ordena la propia lista "in place"
print("Tras sort(reverse=True):", notas)

print("¿Contiene un 9.25?", 9.25 in notas)
print("Suma:", sum(notas), "| Máximo:", max(notas), "| Mínimo:", min(notas))
````

### Recorrer, filtrar y transformar listas

Además del `for` clásico, es habitual usar funciones como `sum()`, `max()`, `min()`, o construir listas
nuevas a partir de otras (esto se ve con más detalle en el apartado de *comprehensions*).


````python
notas = [8.5, 6.0, 9.25, 5.5, 4.0]

aprobados = []
for n in notas:
    if n >= 5:
        aprobados.append(n)
print("Aprobados:", aprobados)

media = sum(notas) / len(notas)
print(f"Media: {media:.2f}")
````

## 7. Tuplas (`tuple`)

Una tupla es como una lista, pero **inmutable**: una vez creada, no se pueden añadir, quitar ni
modificar sus elementos. Se usa para datos que no deben cambiar (coordenadas, un registro fijo,
valores que se devuelven agrupados desde una función...).

> 🔶 **Diferencia con Java:** Java no tiene un tipo tupla en su librería estándar (records de Java 16+
> son lo más parecido). En Python las tuplas son un tipo de datos fundamental y muy usado, entre otras
> cosas para que una función pueda **devolver varios valores a la vez**.


````python
punto = (3, 7)
print(punto, type(punto))
print("x =", punto[0], "| y =", punto[1])

# Desempaquetado: asignar cada elemento de la tupla a una variable en un solo paso
x, y = punto
print(f"x={x}, y={y}")

# Las tuplas son inmutables:
try:
    punto[0] = 99
except TypeError as e:
    print("Error esperado:", e)

# Uso muy habitual: una función que devuelve varios valores a la vez
def estadisticas(numeros):
    return min(numeros), max(numeros), sum(numeros) / len(numeros)

minimo, maximo, media = estadisticas([8.5, 6.0, 9.25, 5.5])
print(f"Mínimo={minimo}, Máximo={maximo}, Media={media:.2f}")
````

## 8. Diccionarios (`dict`)

Un diccionario almacena pares **clave → valor**. Es el equivalente Python de un `HashMap<K, V>` de
Java, pero con una sintaxis mucho más directa y, desde Python 3.7, **manteniendo el orden de
inserción** de las claves.


````python
alumno = {
    "nombre": "Ana García",
    "edad": 19,
    "notas": [8.5, 7.0, 9.0]
}

print(alumno)
print(alumno["nombre"])              # acceso por clave
print(alumno.get("apellido", "Sin especificar"))   # get() con valor por defecto si la clave no existe

alumno["email"] = "ana@example.com"  # añadir una nueva clave
alumno["edad"] = 20                  # modificar un valor existente
print(alumno)

del alumno["email"]                  # eliminar una clave
print(alumno)

print("¿Tiene la clave 'nombre'?", "nombre" in alumno)

print("Claves:", list(alumno.keys()))
print("Valores:", list(alumno.values()))
print("Pares clave-valor:")
for clave, valor in alumno.items(): # [ ['nombre' , 'Ana García' ], ['edad', 19] ... ]
    print(f"  {clave}: {valor}")
````

### Diccionarios anidados (muy útil, por ejemplo, al trabajar con JSON o XML convertido)


````python
alumnos = {
    1: {"nombre": "Ana", "nota": 8.5},
    2: {"nombre": "Luis", "nota": 6.0},
    3: {"nombre": "Marta", "nota": 9.25},
}

for id_alumno, datos in alumnos.items():
    print(f"[{id_alumno}] {datos['nombre']} -> {datos['nota']}")

media_general = sum(d["nota"] for d in alumnos.values()) / len(alumnos)
print(f"\nMedia general: {media_general:.2f}")
````

## 9. Conjuntos (`set`)

Un `set` almacena elementos **sin duplicados** y **sin un orden garantizado**. Es el equivalente
Python de un `HashSet` de Java, y resulta muy cómodo para eliminar duplicados o para operaciones
matemáticas de conjuntos (unión, intersección, diferencia).


````python
asignaturas_ana = {"Programación", "Bases de Datos", "Acceso a Datos"}
asignaturas_luis = {"Acceso a Datos", "Sistemas", "Programación"}

print(asignaturas_ana)

# Eliminar duplicados de una lista es tan sencillo como convertirla a set
numeros_con_duplicados = [1, 2, 2, 3, 3, 3, 4]
print("Sin duplicados:", set(numeros_con_duplicados))

asignaturas_ana.add("Inglés Técnico")
print("Tras add:", asignaturas_ana)

print("Unión:", asignaturas_ana | asignaturas_luis)
print("Intersección (asignaturas comunes):", asignaturas_ana & asignaturas_luis)
print("Diferencia (solo Ana):", asignaturas_ana - asignaturas_luis)
print("¿'Sistemas' está en las de Ana?", "Sistemas" in asignaturas_ana)
````

## 10. Funciones

Se definen con `def`. No es obligatorio declarar el tipo de los parámetros ni del valor devuelto
(aunque se pueden añadir *type hints* opcionales, que no se comprueban en tiempo de ejecución, solo
sirven de documentación y ayudan a herramientas como el IDE).


````python
def saludar(nombre):
    return f"Hola, {nombre}!"

print(saludar("Ana"))

# Parámetros con valor por defecto (equivalente a la sobrecarga de métodos en Java)
def calcular_media(notas, redondeo=2):
    return round(sum(notas) / len(notas), redondeo)

print(calcular_media([8.5, 6.0, 9.25]))
print(calcular_media([8.5, 6.0, 9.25], redondeo=0))

# Parámetros con "type hints" (opcionales, solo documentación, no se comprueban en ejecución)
def es_apto(nota: float, minimo: float = 5.0) -> bool:
    return nota >= minimo

print(es_apto(6.5))
print(es_apto(4.0, minimo=4.5))
````

### `*args` y `**kwargs`: número variable de argumentos

Equivale a los *varargs* de Java (`int... numeros`), pero Python distingue entre argumentos
posicionales variables (`*args`) y argumentos con nombre variables (`**kwargs`).


````python
def sumar_todos(*numeros):          # *args: recoge cualquier cantidad de argumentos en una tupla
    return sum(numeros)

print(sumar_todos(1, 2, 3))
print(sumar_todos(1, 2, 3, 4, 5))

def crear_ficha(**datos):           # **kwargs: recoge argumentos con nombre en un diccionario
    for clave, valor in datos.items():
        print(f"  {clave}: {valor}")

crear_ficha(nombre="Ana", edad=19, curso="2º DAM")
````

### Funciones `lambda` (funciones anónimas)

Equivalen a las expresiones lambda de Java (`(a, b) -> a + b`). Son funciones de **una sola
expresión**, muy usadas como argumento de otras funciones (`sorted()`, `map()`, `filter()`...).


````python
cuadrado = lambda x: x ** 2
print(cuadrado(5))

alumnos = [("Ana", 8.5), ("Luis", 6.0), ("Marta", 9.25)]

# Ordenar una lista de tuplas por el segundo elemento (la nota), usando una lambda como criterio
alumnos_ordenados = sorted(alumnos, key=lambda alumno: alumno[1], reverse=True)
print(alumnos_ordenados)

# map(): aplica una función a cada elemento de un iterable
notas = [8.5, 6.0, 9.25]
notas_redondeadas = list(map(lambda n: round(n), notas))
print(notas_redondeadas)

# filter(): se queda solo con los elementos que cumplen una condición
# Sin lambda
def criterio(alumno):
    nota = alumno[1]
    if nota >= 5:
        return True
    else:
        return False

alumnos_aprobados = list(filter(criterio, alumnos))
print(alumnos_aprobados)

aprobados = list(filter(lambda alumno: alumno[1] >= 5, alumnos))
print(aprobados)
````

## 11. Comprensiones (*comprehensions*)

Son una forma **compacta** de crear listas, diccionarios o conjuntos a partir de otro iterable, en una
sola línea. No existen en Java (lo más parecido serían los *streams* con `map`/`filter`/`collect`),
pero son una de las herramientas más idiomáticas y usadas en Python.

```python
[expresion(elemento) for elemento in iterable if condicion]
```


````python
notas = [8.5, 6.0, 9.25, 5.5, 4.0, 3.0]

# Forma "clásica" con un bucle for
cuadrados = []
for n in range(1, 6):
    cuadrados.append(n ** 2)
print("Con bucle for:", cuadrados)

# La misma lista con una list comprehension
cuadrados_comp = [n ** 2 for n in range(1, 6)]
print("Con comprehension:", cuadrados_comp)

# Comprehension con condición (filtro)
aprobados = [n for n in notas if n >= 5]
print("Aprobados:", aprobados)

# Comprehension con transformación Y condición a la vez
aprobados_redondeados = [round(n) for n in notas if n >= 5]
print("Aprobados redondeados:", aprobados_redondeados)

# Dict comprehension: igual, pero para construir un diccionario
nombres = ["Ana", "Luis", "Marta"]
longitudes = {nombre: len(nombre) for nombre in nombres}
print("Longitudes:", longitudes)

# Set comprehension: igual, pero para construir un conjunto (sin duplicados)
letras_unicas = {letra for palabra in nombres for letra in palabra.lower()}
print("Letras únicas usadas:", letras_unicas)
````

## 12. Manejo básico de errores

Un repaso mínimo, ya que se tratará en profundidad en el cuaderno de manejo de ficheros. La estructura
es `try` / `except` / `else` / `finally`, equivalente al `try`/`catch`/`finally` de Java.


````python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError as exception:
        print(f'Tipo de excepción: {type(exception)}')
        return None
    else:
        # el bloque "else" se ejecuta SOLO si no hubo ninguna excepción (no existe en Java)
        print("División realizada sin problemas")
        return resultado
    finally:
        print("(este mensaje se imprime siempre, haya o no error)")

print(dividir(10, 2))
print()
print(dividir(10, 0))
````

````python
# Manera de obtener el nombre de la excepción
try:
    10 / 0
except Exception as e:
    print(type(e))
````

````python
# Capturar varios tipos de error y acceder al mensaje de la excepción
entradas = ["10", "abc", "0"]

for entrada in entradas:
    try:
        numero = int(entrada)
        resultado = 100 / numero
    # except Exception:
    #     print('Excepción genérica capturada.')
    except ValueError:
        print(f"'{entrada}' no es un número válido")
    except ZeroDivisionError:
        print(f"'{entrada}' es cero, no se puede dividir")
    else:
        print(f"100 / {entrada} = {resultado}")
````

## POO Básico

````python
class Animal:
    nombre_animal = 'Animal'
    def __init__(self, nombre): # Constructor
        self.nombre = nombre

    def mostrar_nombre(self):
        print(f'Nombre del animal: {self.nombre}')

mi_primer_animal = Animal("Pedro")
print(type(mi_primer_animal))
mi_primer_animal.mostrar_nombre()
print(mi_primer_animal.nombre_animal)
````

````python
class Perro(Animal):
    def __init__(self, nombre_animal, nombre_perro, raza):
        super().__init__(nombre_animal)
        self.nombre_perro = nombre_perro
        self.raza = raza
        self._protegido = 'Esto es protegido'
        self.__privado = 'Esto es privado'

    def ladrar(self):
        print('woof woof')

    def mostrar_nombre(self):
        print(f'Nombre del perro: {self.nombre}')

    def mostrar_privado(self):
        print(self.__privado)


mi_perro = Perro('Juan', 'Pepe','Chihuaha')
print(type(mi_perro))
mi_perro.mostrar_nombre()
print(f'Raza: {mi_perro.raza}')
mi_perro.ladrar()
print(f'protegido: {mi_perro._protegido}')
mi_perro.mostrar_privado()
# print(f'privado: {mi_perro.__privado}')
print(mi_perro.nombre_animal)
````

### Ejercicios propuestos

1. Escribe una función `es_primo(n)` que devuelva `True` si `n` es un número primo, usando un bucle
   `for` con `break`/`else`.
2. Dada la lista `notas = [4, 8.5, 6, 9.25, 3, 7, 5]`, calcula con una *list comprehension* la lista de
   notas aprobadas (`>= 5`) redondeadas a un decimal.
3. Crea un diccionario `inventario` donde las claves sean nombres de productos y los valores su
   cantidad en stock. Escribe un bucle que muestre solo los productos con menos de 5 unidades.
4. Escribe una función `estadisticas_texto(texto)` que reciba una cadena y devuelva una **tupla** con
   el número de palabras, el número de caracteres (sin contar espacios) y la palabra más larga.
5. Usa un `set` para averiguar qué letras del abecedario **no** aparecen en la frase
   `"El veloz murciélago hindú comía feliz cardillo y kiwi"` (pista: compara contra
   `set("abcdefghijklmnopqrstuvwxyz")`).
6. Escribe una función `resumen_alumno(**datos)` que reciba un número variable de pares clave-valor
   (por ejemplo `nombre="Ana", nota=8.5, curso="2ºDAM"`) y devuelva una cadena formateada con toda la
   información, usando un f-string.


````python
from math import sqrt

def es_primo(n: int) -> bool:
    for i in range(2, int(sqrt(n)) + 1):
        if n % i == 0:
            return False

    return True

print(es_primo(2*5*7*13))
print(es_primo(4073))
````
