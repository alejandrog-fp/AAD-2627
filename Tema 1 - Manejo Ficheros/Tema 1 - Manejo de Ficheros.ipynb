### Miniproyecto

Crea un programa que solicite al usuario una ruta a una carpeta (tiene que comprobar que la ruta apunte a una carpeta que exista). Imprime por pantalla un resumen de la cantidad de ficheros que hay en esa carpeta, por ejemplo:
```commandline
> c:\carpeta_test
    Carpetas : 3
    Ficheros pdf: 2
    Ficheros docx: 1
```

````python
def resumen_carpeta(ruta_carpeta: Path) -> dict[str, int] | None:

    if ruta_carpeta.exists() and ruta_carpeta.is_dir():
        resumen = dict()

        for elemento in ruta_carpeta.iterdir():

            if elemento.is_dir():
                extension = 'directorio'
            else:
                extension = elemento.suffix[1:]

            try:
                resumen[extension] += 1
            except KeyError:
                resumen[extension] = 1
            # o equivalentemente:
            # if extension in resumen:
            #     resumen['directorio'] += 1
            # else:
            #     resumen['directorio'] = 1

        return resumen

resumen_carpeta(Path(input('> ')))
````

````python
from pathlib import Path

# 1. Obtenemos los datos en bruto
localidades = Path('localidades.csv')

with open(localidades, 'rt', encoding='utf8') as f:
    # cabecera = []
    # for columna in f.readline().split(';'):
    #     cabecera.append(columna.strip())
    cabecera = [columna.strip() for columna in f.readline().split(';')]
    datos = f.readlines()

# 2. Procesamos los datos
datos_procesados = {}

for h in cabecera:
    datos_procesados[h] = []

for localidad in datos:
    localidad_limpio = []
    for atributo in localidad.split(';'):
        try:
            if '.' in atributo:
                localidad_limpio.append(float(atributo.strip()))
            else:
                localidad_limpio.append(int(atributo.strip()))
        except ValueError:
            localidad_limpio.append(atributo.strip())

    for i, c in enumerate(cabecera): # [ [0, cabecera[0] ] , [1 , cabecera[1]] , ...  ]
        datos_procesados[c].append(localidad_limpio[i])

#cabecera = ['Comunidad', 'Provincia']
## 1 iteracion
#i = 0
#c = cabecera[0] = "Comunidad"
#datos_procesados["Comunidad"].append( localidad_limpio[0] )
## 2 iteracion
#i = 1
#c = cabecera[1] = "Provincia"
#datos_procesados["Provincia"].append( localidad_limpio[1] )

# 3. Analizamos los datos y mostramos los resultados
habitantes_max = max(datos_procesados["Total Habitantes"])
habitantes_max_index = datos_procesados["Total Habitantes"].index(habitantes_max)
# Ejercicio: ¿Qué pasaría si dos ciudades tiene el mismo numero de habitantes?

print('¿Cuál es la localidad con mayor número de habitantes?')
print()
print('La localidad con mayor número de habitantes es:')
for k, v in datos_procesados.items():
    print(f'\t{k} : {v[habitantes_max_index]}')
print()

print('¿Cuál es la comunidad con menor número de habitantes?')
print()
comunidades = set(datos_procesados["Comunidad"])
datos_agregados = {comunidad : 0 for comunidad in comunidades}

for comunidad, habitantes in zip(datos_procesados["Comunidad"], datos_procesados["Total Habitantes"]):
    datos_agregados[comunidad] += habitantes
habitantes_comunidad_min = min(datos_agregados.values())
for comunidad, habitantes in datos_agregados.items():
    if habitantes_comunidad_min == habitantes:
        print('La comunidad con menor numero de habitantes es:', comunidad)
        break
````
